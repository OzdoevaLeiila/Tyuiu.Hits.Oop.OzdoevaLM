Создать базовый класс Ticket в C#, который будет представлять билеты на различные мероприятия. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма. Требования к базовому классу Ticket: • Атрибуты: ID билета (TicketId), Номер ряда (RowNumber), Номер места (SeatNumber). • Методы: o PrintTicketInfo(): метод для печати информации о билете. o ReserveSeat(): метод для резервирования места. o GetTicketDetails(): метод для получения деталей билета. Требования к производным классам:

БилетНаКонцерт (ConcertTicket): Должен содержать дополнительные атрибуты, такие как Имя исполнителя (ArtistName). Метод PrintTicketInfo() должен быть переопределен для включения имени исполнителя в информацию о билете. БилетНаСпектакль (PlayTicket): Должен содержать дополнительные атрибуты, такие как Название спектакля (PlayTitle). Метод GetTicketDetails() должен быть переопределен для отображения названия спектакля вместе с другими деталями билета. БилетНаКонференцию (ConferenceTicket) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Название конференции (ConferenceName). Метод ReserveSeat() должен быть переопределен для добавления информации о предпочтениях участника относительно мест.

```
using System;
using System.Collections.Generic;

public class Ticket
{
    public string TicketId { get; set; }
    public int RowNumber { get; set; }
    public int SeatNumber { get; set; }
    public decimal Price { get; set; }
    public bool IsReserved { get; set; }

    public Ticket(string ticketId, int rowNumber, int seatNumber, decimal price)
    {
        TicketId = ticketId;
        RowNumber = rowNumber;
        SeatNumber = seatNumber;
        Price = price;
        IsReserved = false;
    }

    public virtual void PrintTicketInfo()
    {
        Console.WriteLine($"Билет {TicketId}: ряд {RowNumber}, место {SeatNumber}, цена {Price} руб.");
    }

    public virtual void ReserveSeat()
    {
        IsReserved = true;
        Console.WriteLine($"Место {SeatNumber} в ряду {RowNumber} зарезервировано.");
    }

    public virtual string GetTicketDetails()
    {
        return $"ID: {TicketId}, Ряд: {RowNumber}, Место: {SeatNumber}, Цена: {Price}, Зарезервирован: {IsReserved}";
    }

    public void ApplyDiscount(decimal percent)
    {
        Price -= Price * percent / 100;
        Console.WriteLine($"Скидка {percent}% применена. Новая цена: {Price} руб.");
    }

    public void ApplyDiscount(decimal amount, bool isFixed)
    {
        if (isFixed)
        {
            Price -= amount;
            Console.WriteLine($"Фиксированная скидка {amount} руб. Новая цена: {Price} руб.");
        }
        else
        {
            ApplyDiscount(amount);
        }
    }

    public virtual void CancelReservation()
    {
        IsReserved = false;
        Console.WriteLine($"Бронь на билет {TicketId} отменена.");
    }
}

public class ConcertTicket : Ticket
{
    public string ArtistName { get; set; }
    public string Genre { get; set; }

    public ConcertTicket(string ticketId, int rowNumber, int seatNumber, decimal price, string artistName, string genre)
        : base(ticketId, rowNumber, seatNumber, price)
    {
        ArtistName = artistName;
        Genre = genre;
    }

    public override void PrintTicketInfo()
    {
        base.PrintTicketInfo();
        Console.WriteLine($"Исполнитель: {ArtistName}, Жанр: {Genre}");
    }

    public override string GetTicketDetails()
    {
        return base.GetTicketDetails() + $", Исполнитель: {ArtistName}, Жанр: {Genre}";
    }

    public void MeetAndGreet()
    {
        Console.WriteLine($"Доступ к meet-and-greet с {ArtistName}.");
    }
}

public class PlayTicket : Ticket
{
    public string PlayTitle { get; set; }
    public string Theater { get; set; }

    public PlayTicket(string ticketId, int rowNumber, int seatNumber, decimal price, string playTitle, string theater)
        : base(ticketId, rowNumber, seatNumber, price)
    {
        PlayTitle = playTitle;
        Theater = theater;
    }

    public override string GetTicketDetails()
    {
        return base.GetTicketDetails() + $", Спектакль: {PlayTitle}, Театр: {Theater}";
    }

    public override void PrintTicketInfo()
    {
        base.PrintTicketInfo();
        Console.WriteLine($"Спектакль: {PlayTitle}, Театр: {Theater}");
    }

    public void ShowProgram()
    {
        Console.WriteLine($"Программа спектакля '{PlayTitle}' доступна.");
    }
}

public class ConferenceTicket : Ticket
{
    public string ConferenceName { get; set; }
    public string PreferredZone { get; set; }

    public ConferenceTicket(string ticketId, int rowNumber, int seatNumber, decimal price, string conferenceName, string preferredZone)
        : base(ticketId, rowNumber, seatNumber, price)
    {
        ConferenceName = conferenceName;
        PreferredZone = preferredZone;
    }

    public override void ReserveSeat()
    {
        base.ReserveSeat();
        Console.WriteLine($"Участник предпочитает зону: {PreferredZone}.");
    }

    public override void PrintTicketInfo()
    {
        base.PrintTicketInfo();
        Console.WriteLine($"Конференция: {ConferenceName}, Предпочтение: {PreferredZone}");
    }

    public void GetSchedule()
    {
        Console.WriteLine($"Расписание конференции '{ConferenceName}' отправлено на почту.");
    }
}

public class TicketCollection<T> where T : Ticket
{
    private List<T> _tickets = new List<T>();

    public void Add(T ticket) => _tickets.Add(ticket);

    public void PrintAll()
    {
        foreach (var ticket in _tickets)
        {
            ticket.PrintTicketInfo();
            Console.WriteLine();
        }
    }

    public void ReserveAll()
    {
        foreach (var ticket in _tickets)
            ticket.ReserveSeat();
    }
}

TicketCollection<Ticket> allTickets = new TicketCollection<Ticket>();
allTickets.Add(new ConcertTicket("C001", 5, 12, 3500, "Лейла", "Поп"));
allTickets.Add(new PlayTicket("P001", 3, 7, 2000, "Вишнёвый сад", "МХТ"));
allTickets.Add(new ConferenceTicket("K001", 1, 1, 5000, "IT Conf", "Зона A"));

Console.WriteLine("--- Все билеты ---");
allTickets.PrintAll();

Console.WriteLine("--- Резервирование ---");
allTickets.ReserveAll();

Console.WriteLine();

Console.WriteLine("--- Детали ---");
Console.WriteLine(new ConcertTicket("C002", 2, 2, 4000, "Лера", "Рок").GetTicketDetails());
Console.WriteLine(new PlayTicket("P002", 4, 4, 1500, "Гамлет", "БДТ").GetTicketDetails());
Console.WriteLine(new ConferenceTicket("K002", 2, 2, 6000, "Dev Conf", "Зона B").GetTicketDetails());

Console.WriteLine();

Console.WriteLine("--- Перегрузка скидок ---");
Ticket discountTicket = new ConcertTicket("C003", 1, 1, 5000, "Вика", "Джаз");
discountTicket.ApplyDiscount(10);
discountTicket.ApplyDiscount(500, true);
```

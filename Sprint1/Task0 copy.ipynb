----Создайте проект на C#, реализующее классы Товар (Product), Продавец (Seller), Магазин (Store) с использование геттеров и сеттеров, а также модификаторов доступа public, private, protected, internal. От каждого класса создайте по одному экземпляру.

```
using System;

internal class Product
{
    private int _id;
    private string _name;
    private double _price;


    public int Id
    {
        get { return _id; }
        set { _id = value; }
    }

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public double Price
    {
        get { return _price; }
        set { _price = value; }
    }

    public Product(int id, string name, double price)
    {
        Id = id;
        Name = name;
        Price = price;
    }

    public void PrintInfo()
    {
        Console.WriteLine($"Товар: {Name}, цена: {Price}");
    }
}

public class Seller
{
    private string _name;
    protected int _experience;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public int Experience
    {
        get { return _experience; }
        set { _experience = value; }
    }

    public Seller(string name, int experience)
    {
        Name = name;
        Experience = experience;
    }

    public void PrintInfo()
    {
        Console.WriteLine($"Продавец: {Name}, стаж: {Experience} лет");
    }
}

public class Store
{
    private string _storeName;
    private Seller _seller;
    private Product _product;

    public string StoreName
    {
        get { return _storeName; }
        set { _storeName = value; }
    }

    public Seller Seller
    {
        get { return _seller; }
        set { _seller = value; }
    }

    public Product Product
    {
        get { return _product; }
        set { _product = value; }
    }

    public Store(string storeName, Seller seller, Product product)
    {
        StoreName = storeName;
        Seller = seller;
        Product = product;
    }

    public void PrintInfo()
    {
        Console.WriteLine($"Магазин: {StoreName}");
        Seller.PrintInfo();
        Product.PrintInfo();
    }
}

Product myProduct = new Product(1, "Хлеб", 45.5);
Seller mySeller = new Seller("Игорь", 5);
Store myStore = new Store("Пятёрочка", mySeller, myProduct);

Console.WriteLine("--- Товар ---");
myProduct.PrintInfo();

Console.WriteLine("--- Продавец ---");
mySeller.PrintInfo();

Console.WriteLine("--- Магазин ---");
myStore.PrintInfo();
```

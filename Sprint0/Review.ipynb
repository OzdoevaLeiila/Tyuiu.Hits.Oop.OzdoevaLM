Разработать проект для управления растениями в саду, используя основные концепции объектно-ориентированного программирования (ООП) в C#. Вам нужно создать классы, которые будут описывать разные типы растений, используя наследование и полиморфизм.

```
using System;

public class Plant
{
    private string _name;
    private string _type;
    private float _height;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public string Type
    {
        
        get { return _type; }
        set { _type = value; }
    }

    public float Height
    {
        get { return _height; }
        set { _height = value; }
    }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Цветение()
    {
        Console.WriteLine($"Растение {Name} цветёт");
    }

    public void Растет(float рост)
    {
        Height += рост;
        Console.WriteLine($"{Name} вырос на {рост} см. Теперь высота: {Height} см");
    }
}

public class Flower : Plant
{
    private string _color;
    private int _petalsCount;

    public string Color
    {
        get { return _color; }
        set { _color = value; }
    }

    public int PetalsCount
    {
        get { return _petalsCount; }
        set { _petalsCount = value; }
    }

    public Flower(string name, string type, float height, string color, int petalsCount)
        : base(name, type, height)
    {
        Color = color;
        PetalsCount = petalsCount;
    }

    public override void Цветение()
    {
        Console.WriteLine($"Цветок {Name} распускается и радует глаз!");
    }

    public void Увядать()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать...");
    }
}

public class Tree : Plant
{
    private int _age;
    private string _leafType;

    public int Age
    {
        get { return _age; }
        set { _age = value; }
    }

    public string LeafType
    {
        get { return _leafType; }
        set { _leafType = value; }
    }

    public Tree(string name, string type, float height, int age, string leafType)
        : base(name, type, height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Цветение()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями!");
    }

    public void Вырастить(int годы)
    {
        Age += годы;
        Console.WriteLine($"Дерево {Name} выросло на {годы} лет. Новый возраст: {Age} лет");
    }
}

Flower myFlower = new Flower("Роза", "Цветок", 30, "Красный", 20);
Tree myTree = new Tree("Дуб", "Дерево", 500, 10, "Лиственные");

Console.WriteLine("--- Цветок ---");
myFlower.Цветение();
myFlower.Растет(5);
myFlower.Увядать();

Console.WriteLine("--- Дерево ---");
myTree.Цветение();
myTree.Растет(10);
myTree.Вырастить(5);

Console.WriteLine("--- Полиморфизм ---");
Plant[] garden = { myFlower, myTree };
foreach (Plant p in garden)
{
    p.Цветение();
}
```

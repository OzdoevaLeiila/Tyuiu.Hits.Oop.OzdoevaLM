Разработать проект для управления растениями в саду, используя основные концепции объектно-ориентированного программирования (ООП) в C#. Вам нужно создать классы, которые будут описывать разные типы растений, используя наследование и полиморфизм.

```
using System;
using System.Collections.Generic;

public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение [{Name}] цветёт");
    }

    public void Grow(float growth)
    {
        Height += growth;
        Console.WriteLine($"Растение [{Name}] выросло на {growth} см. Новая высота: {Height} см");
    }
}

public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public Flower(string name, float height, string color, int petalCount)
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок [{Name}] распускается и радует глаз!");
        Console.WriteLine($"Цвет: {Color}, лепестков: {PetalCount}");
    }

    public void Wither()
    {
        Console.WriteLine($"Цветок [{Name}] начинает увядать...");
    }
}

public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public Tree(string name, float height, int age, string leafType)
        : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево [{Name}] покрывается новыми листьями!");
        Console.WriteLine($"Возраст: {Age} лет, тип листьев: {LeafType}");
    }

    public void GrowOlder(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево [{Name} стало старше на {years} лет. Новый возраст: {Age} лет");
    }
}

Flower rose = new Flower("Роза", 30f, "Красный", 25);
Tree oak = new Tree("Дуб", 500f, 50, "Лиственные");

Console.WriteLine("=== Цветок ===");
rose.Grow(5f);
rose.Bloom();
rose.Wither();

Console.WriteLine("=== Дерево ===");
oak.Grow(10f);
oak.GrowOlder(2);
oak.Bloom();

Console.WriteLine("=== Полиморфизм ===");
List<Plant> garden = new List<Plant> { rose, oak };
foreach (Plant p in garden)
{
    Console.WriteLine($"Растение: {p.Name} ({p.Type})");
    p.Bloom();
}
```

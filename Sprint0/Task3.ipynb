```
using System;

public class Ticket
{
    public int TicketId;
    public int RowNumber;
    public int SeatNumber;

    public Ticket(int ticketId, int rowNumber, int seatNumber)
    {
        TicketId = ticketId;
        RowNumber = rowNumber;
        SeatNumber = seatNumber;
    }

    public virtual void PrintTicketInfo()
    {
        Console.WriteLine($"Билет №{TicketId}, ряд {RowNumber}, место {SeatNumber}");
    }

    public virtual void ReserveSeat()
    {
        Console.WriteLine($"Место {SeatNumber} зарезервировано");
    }

    public virtual void GetTicketDetails()
    {
        Console.WriteLine($"ID: {TicketId}, ряд: {RowNumber}, место: {SeatNumber}");
    }
}

public class ConcertTicket : Ticket
{
    public string ArtistName;

    public ConcertTicket(int ticketId, int rowNumber, int seatNumber, string artistName)
        : base(ticketId, rowNumber, seatNumber)
    {
        ArtistName = artistName;
    }

    public override void PrintTicketInfo()
    {
        Console.WriteLine($"Билет на концерт №{TicketId}, ряд {RowNumber}, место {SeatNumber}, исполнитель: {ArtistName}");
    }
}

public class PlayTicket : Ticket
{
    public string PlayTitle;

    public PlayTicket(int ticketId, int rowNumber, int seatNumber, string playTitle)
        : base(ticketId, rowNumber, seatNumber)
    {
        PlayTitle = playTitle;
    }

    public override void GetTicketDetails()
    {
        Console.WriteLine($"Билет на спектакль №{TicketId}, ряд {RowNumber}, место {SeatNumber}, спектакль: {PlayTitle}");
    }
}

public class ConferenceTicket : Ticket
{
    public string ConferenceName;

    public ConferenceTicket(int ticketId, int rowNumber, int seatNumber, string conferenceName)
        : base(ticketId, rowNumber, seatNumber)
    {
        ConferenceName = conferenceName;
    }

    public override void ReserveSeat()
    {
        Console.WriteLine($"Место {SeatNumber} зарезервировано для конференции: {ConferenceName} (с учётом предпочтений участника)");
    }
}

ConcertTicket myConcert = new ConcertTicket(1, 5, 10, "Игорь Крутой");
PlayTicket myPlay = new PlayTicket(2, 3, 7, "Гамлет");
ConferenceTicket myConference = new ConferenceTicket(3, 1, 2, "IT-конференция");

Console.WriteLine("--- Концерт ---");
myConcert.PrintTicketInfo();
myConcert.ReserveSeat();
myConcert.GetTicketDetails();

Console.WriteLine("--- Спектакль ---");
myPlay.PrintTicketInfo();
myPlay.ReserveSeat();
myPlay.GetTicketDetails();

Console.WriteLine("--- Конференция ---");
myConference.PrintTicketInfo();
myConference.ReserveSeat();
myConference.GetTicketDetails();
```

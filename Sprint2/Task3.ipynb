<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Ticket в C#, который будет представлять билеты на различные мероприятия. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма. Требования к базовому классу Ticket: • Атрибуты: ID билета (TicketId), Номер ряда (RowNumber), Номер места (SeatNumber). • Методы: o PrintTicketInfo(): метод для печати информации о билете. o ReserveSeat(): метод для резервирования места. o GetTicketDetails(): метод для получения деталей билета. Требования к производным классам:

БилетНаКонцерт (ConcertTicket): Должен содержать дополнительные атрибуты, такие как Имя исполнителя (ArtistName). Метод PrintTicketInfo() должен быть переопределен для включения имени исполнителя в информацию о билете.
БилетНаСпектакль (PlayTicket): Должен содержать дополнительные атрибуты, такие как Название спектакля (PlayTitle). Метод GetTicketDetails() должен быть переопределен для отображения названия спектакля вместе с другими деталями билета.
БилетНаКонференцию (ConferenceTicket) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Название конференции (ConferenceName). Метод ReserveSeat() должен быть переопределен для добавления информации о предпочтениях участника относительно мест.


#### Дополнительное задание
Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;

// ==================== БАЗОВЫЙ КЛАссС ====================
public class Ticket
{
    private int _ticketId;
    private int _rowNumber;
    private int _seatNumber;
    private string _eventDate;
    private decimal _price;
    private bool _isReserved;

    public int TicketId
    {
        get { return _ticketId; }
        set { _ticketId = value; }
    }

    public int RowNumber
    {
        get { return _rowNumber; }
        set { _rowNumber = value; }
    }

    public int SeatNumber
    {
        get { return _seatNumber; }
        set { _seatNumber = value; }
    }

    public string EventDate
    {
        get { return _eventDate; }
        set { _eventDate = value; }
    }

    public decimal Price
    {
        get { return _price; }
        set { _price = value; }
    }

    public bool IsReserved
    {
        get { return _isReserved; }
        set { _isReserved = value; }
    }

    public Ticket(int ticketId, int rowNumber, int seatNumber, string eventDate, decimal price)
    {
        TicketId = ticketId;
        RowNumber = rowNumber;
        SeatNumber = seatNumber;
        EventDate = eventDate;
        Price = price;
        IsReserved = false;
    }

    public virtual void PrintTicketInfo()
    {
        Console.WriteLine($"Билет №{TicketId}, ряд {RowNumber}, место {SeatNumber}, дата: {EventDate}");
    }

    public virtual void ReserveSeat()
    {
        IsReserved = true;
        Console.WriteLine($"Место {SeatNumber} зарезервировано");
    }

    public virtual void GetTicketDetails()
    {
        Console.WriteLine($"ID: {TicketId}, ряд: {RowNumber}, место: {SeatNumber}, цена: {Price} руб.");
    }

    public virtual decimal CalculateTotal()
    {
        return Price;
    }

    public void ShowStatus()
    {
        Console.WriteLine(IsReserved ? "Статус: забронирован" : "Статус: свободен");
    }

    public void ApplyDiscount(decimal percent)
    {
        Price = Price - (Price * percent / 100);
        Console.WriteLine($"Применена скидка {percent}%. Новая цена: {Price} руб.");
    }
}

// ==================== ИНТЕРФЕЙСЫ (множественное наследование) ====================
public interface IAgeRestricted
{
    int MinAge { get; set; }
    void CheckAge(int age);
}

public interface IOnlineAccess
{
    string StreamLink { get; set; }
    void ProvideOnlineAccess();
}

public interface IVipAccess
{
    void OpenVipLounge();
}

// ==================== ПРОСТОЕ НАСЛЕДОВАНИЕ ====================
public class ConcertTicket : Ticket
{
    private string _artistName;
    private string _genre;

    public string ArtistName
    {
        get { return _artistName; }
        set { _artistName = value; }
    }

    public string Genre
    {
        get { return _genre; }
        set { _genre = value; }
    }

    public ConcertTicket(int ticketId, int rowNumber, int seatNumber, string eventDate, decimal price,
                         string artistName, string genre)
        : base(ticketId, rowNumber, seatNumber, eventDate, price)
    {
        ArtistName = artistName;
        Genre = genre;
    }

    public override void PrintTicketInfo()
    {
        Console.WriteLine($"Билет на концерт №{TicketId}, ряд {RowNumber}, место {SeatNumber}, исполнитель: {ArtistName}, жанр: {Genre}");
    }

    public override decimal CalculateTotal()
    {
        return Price + 500;
    }

    public void MeetAndGreet()
    {
        Console.WriteLine($"Встреча с {ArtistName} после концерта!");
    }
}

public class PlayTicket : Ticket
{
    private string _playTitle;
    private string _director;

    public string PlayTitle
    {
        get { return _playTitle; }
        set { _playTitle = value; }
    }

    public string Director
    {
        get { return _director; }
        set { _director = value; }
    }

    public PlayTicket(int ticketId, int rowNumber, int seatNumber, string eventDate, decimal price,
                      string playTitle, string director)
        : base(ticketId, rowNumber, seatNumber, eventDate, price)
    {
        PlayTitle = playTitle;
        Director = director;
    }

    public override void GetTicketDetails()
    {
        Console.WriteLine($"Билет на спектакль №{TicketId}, ряд: {RowNumber}, место: {SeatNumber}, спектакль: {PlayTitle}, режиссёр: {Director}");
    }

    public void ShowProgram()
    {
        Console.WriteLine($"Программа спектакля '{PlayTitle}'");
    }
}

// ==================== СЛОЖНОЕ НАСЛЕДОВАНИЕ ====================
public class VipConcertTicket : ConcertTicket, IVipAccess
{
    private string _vipZone;

    public string VipZone
    {
        get { return _vipZone; }
        set { _vipZone = value; }
    }

    public VipConcertTicket(int ticketId, int rowNumber, int seatNumber, string eventDate, decimal price,
                            string artistName, string genre, string vipZone)
        : base(ticketId, rowNumber, seatNumber, eventDate, price, artistName, genre)
    {
        VipZone = vipZone;
    }

    public override void PrintTicketInfo()
    {
        base.PrintTicketInfo();
        Console.WriteLine($"VIP-зона: {VipZone}");
    }

    public override decimal CalculateTotal()
    {
        return base.CalculateTotal() + 3000;
    }

    public void OpenVipLounge()
    {
        Console.WriteLine($"Доступ в VIP-лаунж зоны {VipZone} открыт.");
    }
}

// ==================== МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ (через интерфейсы) ====================
public class ConferenceTicket : Ticket, IAgeRestricted, IOnlineAccess
{
    private string _conferenceName;
    private int _minAge;
    private string _streamLink;

    public string ConferenceName
    {
        get { return _conferenceName; }
        set { _conferenceName = value; }
    }

    public int MinAge
    {
        get { return _minAge; }
        set { _minAge = value; }
    }

    public string StreamLink
    {
        get { return _streamLink; }
        set { _streamLink = value; }
    }

    public ConferenceTicket(int ticketId, int rowNumber, int seatNumber, string eventDate, decimal price,
                            string conferenceName, int minAge, string streamLink)
        : base(ticketId, rowNumber, seatNumber, eventDate, price)
    {
        ConferenceName = conferenceName;
        MinAge = minAge;
        StreamLink = streamLink;
    }

    public override void ReserveSeat()
    {
        IsReserved = true;
        Console.WriteLine($"Место {SeatNumber} зарезервировано для конференции: {ConferenceName} (с учётом предпочтений участника)");
    }

    public void CheckAge(int age)
    {
        if (age < MinAge)
            Console.WriteLine($"Доступ запрещён. Минимальный возраст: {MinAge}");
        else
            Console.WriteLine($"Доступ разрешён. Возраст {age} соответствует требованиям.");
    }

    public void ProvideOnlineAccess()
    {
        Console.WriteLine($"Онлайн-доступ предоставлен: {StreamLink}");
    }
}

// ==================== ИСПОЛЬЗОВАНИЕ ====================
ConcertTicket myConcert = new ConcertTicket(1, 5, 10, "15.06.2025", 3000, "Игорь Крутой", "Эстрада");
PlayTicket myPlay = new PlayTicket(2, 3, 7, "20.06.2025", 2500, "Гамлет", "Иван Петров");
VipConcertTicket myVip = new VipConcertTicket(4, 1, 1, "15.06.2025", 8000, "Игорь Крутой", "Эстрада", "A1");
ConferenceTicket myConference = new ConferenceTicket(3, 1, 2, "25.06.2025", 5000, "IT-конференция", 18, "https://stream.itconf.ru");

Console.WriteLine("--- Концерт ---");
myConcert.PrintTicketInfo();
myConcert.ReserveSeat();
myConcert.GetTicketDetails();
myConcert.MeetAndGreet();
myConcert.ShowStatus();
myConcert.ApplyDiscount(10);
Console.WriteLine($"Итоговая стоимость: {myConcert.CalculateTotal()} руб.");

Console.WriteLine("\n--- Спектакль ---");
myPlay.PrintTicketInfo();
myPlay.ReserveSeat();
myPlay.GetTicketDetails();
myPlay.ShowProgram();
myPlay.ShowStatus();
myPlay.ApplyDiscount(5);
Console.WriteLine($"Итоговая стоимость: {myPlay.CalculateTotal()} руб.");

Console.WriteLine("\n--- VIP-концерт ---");
myVip.PrintTicketInfo();
myVip.ReserveSeat();
myVip.GetTicketDetails();
myVip.MeetAndGreet();
myVip.OpenVipLounge();
myVip.ShowStatus();
myVip.ApplyDiscount(15);
Console.WriteLine($"Итоговая стоимость: {myVip.CalculateTotal()} руб.");

Console.WriteLine("\n--- Конференция ---");
myConference.PrintTicketInfo();
myConference.ReserveSeat();
myConference.GetTicketDetails();
myConference.CheckAge(20);
myConference.CheckAge(16);
myConference.ProvideOnlineAccess();
myConference.ShowStatus();
myConference.ApplyDiscount(20);
Console.WriteLine($"Итоговая стоимость: {myConference.CalculateTotal()} руб.");

Console.WriteLine("\n--- Взаимодействие объектов ---");
myPlay.TicketId = myConcert.TicketId;
Console.WriteLine($"ID билета на спектакль изменён на ID концерта: {myPlay.TicketId}");
```
